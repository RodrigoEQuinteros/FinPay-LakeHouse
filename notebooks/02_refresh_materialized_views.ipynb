```python
%sql
-- ============================================================
-- 02_refresh_materialized_views
-- Actualiza los datos del modelo dimensional
-- Ejecutado por finpay_semantic_job.yml tras terminar el ETL
-- ============================================================

USE CATALOG fintech_finpay;

-- 1. Refrescar Dimensiones (Independientes)
REFRESH MATERIALIZED VIEW gold.dim_channel;
REFRESH MATERIALIZED VIEW gold.dim_date;
REFRESH MATERIALIZED VIEW gold.dim_merchant;
REFRESH MATERIALIZED VIEW gold.dim_user;

-- 2. Refrescar Tabla de Hechos (Depende de las dimensiones)
REFRESH MATERIALIZED VIEW gold.fact_transactions;

-- ============================================================
-- 3. Verificación post-refresh para los logs del Job
-- ============================================================
SELECT 'fact_transactions' AS vista, COUNT(*) AS total_filas FROM gold.fact_transactions
UNION ALL
SELECT 'dim_merchant',  COUNT(*) FROM gold.dim_merchant
UNION ALL
SELECT 'dim_user',      COUNT(*) FROM gold.dim_user
UNION ALL
SELECT 'dim_channel',   COUNT(*) FROM gold.dim_channel
UNION ALL
SELECT 'dim_date',      COUNT(*) FROM gold.dim_date
ORDER BY vista;
```
